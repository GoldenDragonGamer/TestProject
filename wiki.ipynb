```python
import requests
import time
from bs4 import BeautifulSoup

list_of_links = []

first_link = ""
def wiki(link):
    global list_of_links
    global first_link
    response = requests.get(
        url=link,
    )

    soup = BeautifulSoup(response.content, 'html.parser')
    ps = soup.find(id="bodyContent").find_all('p')
    for p in ps:
        links = p.find_all("a")
        for link in links:
            if link['href'].find("/wiki/") == -1:
                continue

            first_link = link
            break
    print("https://en.wikipedia.org"+first_link['href'])
    if list_of_links.__contains__(first_link['href']):
        print("End of path")
        exit
    else:
        list_of_links.append(first_link['href'])
        time.sleep(0.1)
        wiki("https://en.wikipedia.org"+first_link['href'])


prev_link = "testtesttest"
wiki("https://en.wikipedia.org/wiki/Computational_thinking")
```
